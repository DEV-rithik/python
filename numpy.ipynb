```python
#creating a numpy array
import numpy as np
print(np.__version__)
a = np.array([1,2,3,4], dtype=np.int32)
print(a.shape)
print(a.ndim)
print(a)
```

```python
#creating different type of arrays using numpy

zeros = np.zeros((3,4))
print("zero matrix :\n",zeros)

ones = np.ones((2,3))
print("ones matrix :\n",ones)

a = np.full((5,5),15)
print("matrix containing 15 :\n",a)

random = np.random.random((2,2))
print(random)
```
